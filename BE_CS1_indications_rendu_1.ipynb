Cette dernière approche semble plus robuste au regard de la régularité du fichier d'entrée. Identifiez le délimiteur qui permet de séparer les différentes actions de l'utilisateur et utilisez-le pour construire la liste des actions réalisées.

```python
import math                       #utilisé par get_grid
import numpy as np                
import pandas as pd
import matplotlib.pyplot as plt   #utilisé par get_grid et les graphiques

def read_ds(ds_name: str):
    with open(f'{ds_name}.csv', encoding='utf-8') as f:     #encoding='utf-8' 
        max_actions = max(( len( str(c).split(",")) for c in f.readlines() ))
        f.seek(0)
        _names = ["util", "navigateur"] if "train" in ds_name else ["navigateur"]
        _names.extend(range(max_actions - len(_names)))
        return pd.read_csv(f, names=_names, dtype= str)
```

Quelques fonctions « utilitaires » :

```python
import warnings
from IPython.display import display, Markdown


# affiche un DataFrame Pandas sous forme de tableau Markdown dans un notebook Jupyter.
def markdown_table(headNtail=False, use_index=True, title=None, precision=2):
    def _get_value(val): return str(round(val, precision) if isinstance(val, float) else val)
    def _format_row(row): 
        row_str = ""
        if use_index: row_str += f"|{str(row.name)}"
        for value in row.values: row_str += f"| {_get_value(value)}"
        return row_str + "|"
    def _get_str(df):
        return "\n".join(df.apply(_format_row, axis=1))
    def _deco(f):
        def _f(*args, **kwargs):
            df = f(*args, **kwargs)
            _str = f"#### {title}\n" if title else ""
            header = ([str(df.index.name)] if use_index else []) + df.columns.astype(str).to_list() 
            _str += f"|{'|'.join(header)}|" + f"\n|{'--|'*len(header)}\n" if header else None
            if headNtail:
                _str += _get_str(df.head())
                _str += "\n|...|...|\n"
                _str += _get_str(df.tail())
            else:
                _str += _get_str(df)
            display(Markdown(_str))
        return _f
    return _deco

# fonction utilitaire permettant d'obtenir une grille graphique à partir d'un nombre arbitraire de lignes/colonnes ou de données.
def get_grid(n, n_row=None, n_col=None, titles=None, figsize=(10, 8), wspace=.5, hspace=.5, **kwargs):
    if n_row: n_col= n_col or math.floor(n/n_row)
    elif n_col: n_row= n_row or math.ceil(n/n_col)
    else:
        n_row = math.ceil(math.sqrt(n))
        n_col = math.floor(n/n_row)
    fig, axs = plt.subplots(n_row, n_col, figsize=figsize, **kwargs)
    plt.subplots_adjust(hspace=hspace, wspace=wspace)
    if titles is not None:
        for ax, title in zip(axs.flat, titles): ax.set_title(title)
    return fig, axs
```

## Construction de caractéristiques

### Traitements préliminaires

En fonction du modèle choisi, différentes étapes de pré-traitement peuvent être nécessaire. Voir https://scikit-learn.org/stable/supervised_learning.html pour des exemples de modèles dans le cas supervisé (notre cas). Quelques exemples :

- Modèle ensembliste (Random Forest, XGBoost, etc.)
- Réseaux de neurones 

Il faut commencer par se demander s'il y a des valeurs aberrantes (*outliers*) et, le cas échéant, appliquer le traitement approprié (pour le moment vous pouvez à priori les supprimer).

```python
# Valeurs aberrantes : critère de Tukey (1,5 x IQR) sur la longueur des sessions
def iqr_bounds(s, k=1.5):
    q1, q3 = s.quantile([0.25, 0.75])
    return q1 - k * (q3 - q1), q3 + k * (q3 - q1)

long_train, long_test = acts_train.groupby(level=0).size(), acts_test.groupby(level=0).size()
low, high = iqr_bounds(long_train)
print(f"Sessions 'trop longues' : train {(long_train > high).mean():.1%}, test {(long_test > high).mean():.1%}")
print("Utilisateurs avec moins de 5 sessions :", (features_train["util"].value_counts() < 5).sum())
```

**Décision.** Environ 9 % des sessions sont « trop longues » selon le critère de Tukey, mais ce sont de vraies sessions (restées ouvertes longtemps), et **le test en contient la même proportion**. Les supprimer reviendrait à apprendre sur des données différentes de celles du test, et à jeter les sessions les plus riches en information. On garde donc toutes les lignes, et on utilise des caractéristiques insensibles à la longueur (fréquences, TF-IDF).

### Variable de classe
Notre variable dépendante est une chaîne de caractères (str). Nous pouvons la convertir en codes catégoriels (numériques) à l'aide de la fonction pd.Categorical.

```python
features_train["util"] = pd.Categorical(features_train["util"])
```

pd.Categorical ne modifie pas directement l'ID utilisateur en un nombre, mais lui ajoute un attribut cat.codes. Nous pouvons créer une petite fonction pour convertir la variable de classe d'une chaîne de caractères en son ID de catégorie :

```python
def to_categories(df, col="util"):
    df[[col]] = df[[col]].apply(lambda x: x.cat.codes)
```

### Obtention des caractéristiques...

Nous allons maintenant créer des caractéristiques à partir de l'ensemble de données. Pour cela, il va falloir comprendre leur format, les parser et voir comment les agréger. On peut commencer par inspecter toutes les actions différentes, indépendemment des informations sur l'écran ou la fiche travaillée.

```python
def filter_action(value: str):
    for delim in [ "(", "<", "$", "1"]:
        if delim in value and (low_ind := value.index(delim)):
            value = value[:low_ind]
    return value

uniques = features_train.iloc[:,2:].stack().dropna().unique()     
filtered_uniques = list(set([filter_action(un) for un in uniques if not un.startswith("t")]))
len(filtered_uniques), filtered_uniques
```

```python
vocab = sorted(filtered_uniques)   # les 35 types d'actions, appris sur le train

def build_features(df, acts):
    """Caractéristiques d'une session : volume, durée, rythme, fréquence de chaque type d'action."""
    actions, temps = acts[~is_time(acts)], acts[is_time(acts)]
    feats = pd.DataFrame(index=df.index)
    feats["n_actions"] = actions.groupby(level=0).size().reindex(df.index, fill_value=0)
    feats["n_actions_distinctes"] = actions.groupby(level=0).nunique().reindex(df.index, fill_value=0)
    feats["duree_s"] = temps.str[1:].astype(int).groupby(level=0).max().reindex(df.index, fill_value=0)
    feats["actions_means"] = feats["n_actions"] / (temps.groupby(level=0).size().reindex(df.index, fill_value=0) + 1)
    # fréquences relatives (et non des comptes) : une session longue et une courte du même utilisateur se ressemblent
    counts = (actions.map(filter_action).groupby(level=0).value_counts().unstack(fill_value=0)
              .reindex(index=df.index, columns=vocab, fill_value=0))
    freqs = counts.div(counts.sum(axis=1).replace(0, 1), axis=0).add_prefix("freq_")
    feats["navigateur"] = df["navigateur"]
    if "util" in df.columns:
        feats["util"] = df["util"]
    return pd.concat([feats, freqs], axis=1)

train_df = build_features(features_train, acts_train)
to_categories(train_df)         

# Les actions sont maintenant dans acts_train / acts_test : on libère le tableau large (plusieurs Go)
features_train = features_train[["util", "navigateur"]].copy()
features_test = features_test[["navigateur"]].copy()
train_df.iloc[:3, :8]
```

Ensuite, on peut employer des fonctions plus complexes pour extraire l'écran le plus utilisé, la configuration d'écran la plus utilisé et la chaîne (catégorie) de la fiche la plus utilisé. Pour vous aider, voici les expressions régulières (*regex*) qui permettent d'extraire de telles informations. N'hésitez pas à extraire d'autres informations si cela vous parait pertinent.

```python
import re
from collections import Counter

pattern_ecran = re.compile(r"\((.*?)\)")
pattern_conf_ecran = re.compile(r"<(.*?)>")
pattern_chaine = re.compile(r"\$(.*?)\$")

def extra_features(df, acts):
    """Pour l'écran, la configuration d'écran et la chaîne : nombre de valeurs distinctes et part des actions qui en ont une."""
    acts = acts[~is_time(acts)]
    out = pd.DataFrame(index=df.index)
    for nom, pat in [("ecran", pattern_ecran), ("conf", pattern_conf_ecran), ("chaine", pattern_chaine)]:
        vals = acts.str.extract(pat, expand=False)
        out[f"n_{nom}_distincts"] = vals.groupby(level=0).nunique().reindex(df.index, fill_value=0)
        out[f"part_{nom}"] = vals.notna().groupby(level=0).mean().reindex(df.index, fill_value=0)
    return out

train_df = pd.concat([train_df, extra_features(features_train, acts_train)], axis=1)
```

### Traitement des chaînes de caractère
La colonne navigateur ne peut prendre que quatre valeurs ; pour la convertir en nombre, on a deux choix : 
- la convertir en variable catégorielle comme nous l'avons fait avec notre variable dépendante
- la convertir en One-Hot Encoding (OHE), un pour chaque navigateur. Pandas propose également une fonction pour cela : get_dummies

En fonction du modèle, l'un ou l'autre peut être plus pertinent (par exemple, pour un arbre de décision ?).

```python
# navigateur : 4 valeurs -> One-Hot 
# pour un arbre de décision, les deux conviendraient.
NAVIGATEURS = sorted(features_train["navigateur"].unique())

def encode_strings(df):
    nav = pd.Categorical(df["navigateur"], categories=NAVIGATEURS)
    return pd.concat([df.drop(columns="navigateur"),
                      pd.get_dummies(nav, prefix="nav", dtype=int).set_axis(df.index)], axis=1)

train_df = encode_strings(train_df)
X_tab, y = train_df.drop(columns="util"), train_df["util"].to_numpy()
print("Caractéristiques tabulaires :", X_tab.shape)
```

## Développement d'une solution de classification automatique

Une fois que vous avez exploré les données et mieux compris comment résoudre la tâche qui vous est confiée, c'est le moment de commencer à développer votre solution de classification automatique supervisée.

Le jeu de validation vous aidera dans la sélection du meilleur modèle. Suivant les algorithmes employés, et leur coût computationnel, vous pourriez aussi avoir recours à la validation croisée à K-folds (*K-fold cross validation*).

- Arbres de décisions simples
- Ensembles d'arbres (Random Forest, XGBoost) : nombre d'arbres
- Réseaux de neurones artificiels (simple MLP) : nombre de couches, nombre de neurones par couche

Vous pouvez également tester les méthodes de régularisation afin d'obtenir des modèles parcimonieux.

Pour cela, vous privilégierez l'usage de la librairie *scikit-learn*.

### Deux représentations des sessions

- **(A) Caractéristiques tabulaires** construites plus haut (volume, durée, rythme, fréquences des 35 types d'actions, écrans / configurations / chaînes, navigateur).
- **(B) La session vue comme un texte** : chaque action est un « mot », et on construit un sac de mots pondéré en **TF-IDF**. Un mot fréquent dans la session mais rare dans les autres sessions (un écran utilisé par un seul utilisateur) pèse beaucoup ; un mot présent partout pèse peu. On utilise cinq familles de mots : actions complètes, types d'actions, écrans / configurations / chaînes, **paires d'actions consécutives** et **triplets de types consécutifs** (ils capturent l'ordre, donc les habitudes de navigation). Les marqueurs de temps sont retirés.

Avec ~57 000 dimensions creuses pour ~3 300 sessions, un **SVM linéaire** est adapté : rapide, robuste en grande dimension et interprétable (un poids par mot). Le TF-IDF est placé dans un `Pipeline`, pour être appris uniquement sur les données d'apprentissage de chaque pli (pas de fuite d'information).

```python
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, top_k_accuracy_score
warnings.filterwarnings("ignore", category=UserWarning)   # « classe peu peuplée » en validation croisée

def sessions_from(df, acts):
    """Une ligne par session : la liste ordonnée de ses actions (sans marqueurs de temps) et le navigateur."""
    seqs = acts[~is_time(acts)].groupby(level=0).agg(list).reindex(df.index)
    return pd.DataFrame({"seq": seqs.apply(lambda v: v if isinstance(v, list) else []),
                         "navigateur": df["navigateur"]}, index=df.index)

def ngrams(seq, n): return ["|".join(seq[i:i + n]) for i in range(len(seq) - n + 1)]
def tok_actions(seq): return seq
def tok_types(seq): return [filter_action(a) for a in seq]
def tok_meta(seq):
    return ([f"E:{m}" for a in seq for m in pattern_ecran.findall(a)] + [f"C:{m}" for a in seq for m in pattern_conf_ecran.findall(a)]
            + [f"H:{m}" for a in seq for m in pattern_chaine.findall(a)])
def tok_bigrammes(seq): return ngrams(seq, 2)
def tok_trigrammes(seq): return ngrams(tok_types(seq), 3)

def pipeline_tfidf(C=1.0):
    blocs = [(f.__name__, TfidfVectorizer(analyzer=f, min_df=m, sublinear_tf=True), "seq")
             for f, m in [(tok_actions, 2), (tok_types, 1), (tok_meta, 2), (tok_bigrammes, 2), (tok_trigrammes, 2)]]
    blocs.append(("navigateur", OneHotEncoder(handle_unknown="ignore"), ["navigateur"]))
    return Pipeline([("features", ColumnTransformer(blocs)), ("svm", LinearSVC(C=C))])

S_train = sessions_from(features_train, acts_train)
```

### Comparaison par validation croisée

**Validation croisée stratifiée à 5 plis** (stratifiée car certains utilisateurs n'ont que 4 sessions). Métriques : **F1 macro**, qui correspond au **F1 score moyen** du challenge (moyenne des F1 de chaque utilisateur, avec F1 = 2pr / (p + r), p la précision et r le rappel : chaque utilisateur pèse autant, quel que soit son nombre de sessions), et **accuracy**, plus intuitive. On compare plusieurs familles de modèles sur chacune des deux représentations : régression logistique, SVM à noyau, arbre de décision, forêts aléatoires, gradient boosting et réseau de neurones sur la représentation A ; SVM linéaire (pour plusieurs valeurs de la régularisation C), Passive-Aggressive, k plus proches voisins et Naive Bayes sur la représentation B.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import ExtraTreesClassifier, HistGradientBoostingClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import ComplementNB

RANDOM_STATE = 42
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

def score_cv(modele, X):
    r = cross_validate(modele, X, y, cv=cv, scoring=["accuracy", "f1_macro"], n_jobs=2)
    return {"accuracy": r["test_accuracy"].mean(), "± écart-type": r["test_accuracy"].std(),
            "F1 macro": r["test_f1_macro"].mean(), "± écart-type F1": r["test_f1_macro"].std()}

# (A) Modèles sur les caractéristiques tabulaires. Ceux qui reposent sur des distances ou des produits
#     scalaires (régression logistique, SVM, MLP) sont précédés d'une standardisation ; les arbres n'en ont pas besoin.
modeles_A = {
    "(A) Régression logistique":        make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000)),
    "(A) SVM noyau RBF":                make_pipeline(StandardScaler(), SVC(kernel="rbf", C=10)),
    "(A) Arbre de décision":            DecisionTreeClassifier(random_state=RANDOM_STATE),
    "(A) Random Forest, 200 arbres":    RandomForestClassifier(200, random_state=RANDOM_STATE),
    "(A) Extra Trees, 200 arbres":      ExtraTreesClassifier(200, random_state=RANDOM_STATE),
    "(A) Gradient boosting":            HistGradientBoostingClassifier(max_iter=200, learning_rate=0.05, max_leaf_nodes=15,
                                                                       min_samples_leaf=5, l2_regularization=1.0,
                                                                       early_stopping=False, random_state=RANDOM_STATE),
    "(A) Réseau de neurones (MLP 256)": make_pipeline(StandardScaler(), MLPClassifier((256,), max_iter=500, random_state=RANDOM_STATE)),
}

# (B) Modèles sur la représentation TF-IDF (même pipeline, seul le classifieur change)
def pipeline_avec(clf):
    p = pipeline_tfidf()
    p.steps[-1] = ("clf", clf)
    return p

modeles_B = {
    "(B) kNN cosinus, k=1":           pipeline_avec(KNeighborsClassifier(1, metric="cosine")),
    "(B) Naive Bayes complémentaire": pipeline_avec(ComplementNB(alpha=0.1)),
    "(B) Passive-Aggressive":         pipeline_avec(SGDClassifier(loss="hinge", penalty=None, learning_rate="pa1",
                                                                  eta0=1.0, random_state=RANDOM_STATE)),
}
modeles_B.update({f"(B) TF-IDF + SVM linéaire, C={C}": pipeline_tfidf(C) for C in [0.1, 1, 3]})

resultats = {nom: score_cv(m, X_tab) for nom, m in modeles_A.items()}
resultats.update({nom: score_cv(m, S_train) for nom, m in modeles_B.items()})
tableau = pd.DataFrame(resultats).T.sort_values("accuracy", ascending=False)
display(tableau.round(4))
BEST_C = float(tableau.filter(like="SVM linéaire", axis=0)["accuracy"].idxmax().split("=")[1])
print("Modèle retenu : TF-IDF + SVM linéaire, C =", BEST_C)
```

**Analyse.** Le SVM linéaire sur la représentation TF-IDF (≈ **0,969** de F1 macro, 0,977 d'accuracy) obtient le meilleur score sur les deux métriques ; le Passive-Aggressive obtient le même F1 macro, à 0,0001 près. Tous les modèles linéaires sur TF-IDF dépassent tous les modèles sur les caractéristiques tabulaires, dont le meilleur est Extra Trees (≈ 0,843 de F1 macro, 0,873 d'accuracy). L'écart, environ 10 points, est bien supérieur à l'écart-type entre plis (≈ 0,5 à 1 point) : il est significatif.
- **La représentation compte plus que le modèle** : les variables agrégées perdent les écrans précis utilisés et l'ordre des actions, très propres à chaque utilisateur. Seul le Naive Bayes (≈ 0,852) fait exception sur B, car il suppose les mots indépendants alors que les n-grammes se recouvrent.
- **Sur la représentation A**, les ensembles d'arbres sont les meilleurs ; l'arbre seul sur-apprend (≈ 0,675) et le gradient boosting (≈ 0,746) souffre du très faible nombre d'exemples par classe.
- **Régularisation** : C = 0,1 régularise trop (0,943 de F1 macro, 0,962 d'accuracy) ; plateau à partir de C = 1. On retient **C = 3**, le meilleur sur les deux métriques.

### Interprétabilité avec les modèles de ML
Nous prenons pour example les modèles de forêt aléatoire, mais ces questions peuvent aussi être applicables dans le cas d'autres modèles.

La forêt aléatoire devrait nous permettre de répondre à cinq questions :

* Quel est notre degré de confiance dans nos prédictions à partir d'une ligne de données particulière ?
* Pour prédire à partir d'une ligne de données particulière, quels ont été les facteurs les plus importants et comment ont-ils influencé cette prédiction ?
* Quelles colonnes sont les prédicteurs les plus forts ?
* Quelles colonnes sont effectivement redondantes les unes par rapport aux autres, aux fins de la prédiction ?
* Comment les prédictions varient-elles lorsque nous modifions ces colonnes ?

```python
# Forêt aléatoire (représentation A) apprise sur la même partie train
rf = RandomForestClassifier(200, random_state=RANDOM_STATE, n_jobs=2).fit(X_tab.iloc[idx_tr], y[idx_tr])
proba = rf.predict_proba(X_tab.iloc[idx_va])

# 1. Confiance : la probabilité maximale est-elle liée à la justesse ?
juste = rf.classes_[proba.argmax(axis=1)] == y[idx_va]
display(pd.Series(juste).groupby(pd.cut(proba.max(axis=1), [0, .2, .4, .6, 1]), observed=True).mean()
        .round(3).to_frame("accuracy par tranche de confiance"))

# 3. Prédicteurs les plus forts (importance MDI)
pd.Series(rf.feature_importances_, index=X_tab.columns).sort_values()[-15:].plot.barh(figsize=(8, 5), title="Importance des variables (Random Forest)")
plt.show()

# 4. Redondance : paires de variables très corrélées
corr = X_tab.corr(method="spearman").abs()
print("Paires |rho| > 0,9 :", [(a, b) for i, a in enumerate(corr) for b in corr.columns[i + 1:] if corr.loc[a, b] > 0.9])
```

```python
# 2. Facteurs d'une prédiction particulière (SVM linéaire : score = somme des poids x valeurs TF-IDF)
feats, svm = modele.named_steps["features"], modele.named_steps["svm"]
x = feats.transform(S_train.iloc[[idx_va[0]]])
c = svm.predict(x)[0]
contrib = x.multiply(svm.coef_[list(svm.classes_).index(c)]).toarray().ravel()
top = np.argsort(contrib)[::-1][:8]
print(f"Session prédite : classe {c} (vraie : {y[idx_va[0]]})")
pd.DataFrame({"mot": [feats.get_feature_names_out()[t][:90] for t in top], "contribution": contrib[top].round(3)})
```

**Réponses aux cinq questions.**
1. **Confiance** : la probabilité maximale de la forêt est informative. Au-dessus de 0,4, les prédictions sont justes à 98–100 % ; en dessous de 0,2, à moins de 50 %.
2. **Facteurs d'une prédiction** : pour le SVM, chaque terme poids × valeur TF-IDF est la contribution exacte d'un mot. Ici, la session est attribuée à son utilisateur surtout grâce à une configuration d'écran très spécifique (`MT_AppelN1`) et à l'usage de la grille d'historique de recherche.
3. **Prédicteurs les plus forts** : les fréquences de raccourcis, de clics sur la grille d'historique de recherche, de double-clics, de retours sur un écran et de filtrage / tri. Ce sont des **habitudes d'utilisation** plus que des volumes.
4. **Redondance** : `n_actions` et `duree_s` sont quasi équivalentes (une longue session dure longtemps), de même que les ouvertures et fermetures de dialogues, ou les sélections et désélections d'éléments. On pourrait garder une variable par paire sans perte.
5. **Effet d'une variable** : il se mesure par la dépendance partielle (`sklearn.inspection.PartialDependenceDisplay`), qui fait varier une variable et suit la probabilité prédite ; par exemple, la probabilité d'un utilisateur « clavier » augmente avec sa fréquence de raccourcis.
